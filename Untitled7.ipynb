```python
from google.analytics.data_v1beta import BetaAnalyticsDataClient
from google.analytics.data_v1beta.types import RunReportRequest, DateRange, Dimension, Metric
from google.cloud import bigquery
from google.oauth2 import service_account
import pandas as pd
from datetime import datetime

# === CONFIG ===
GA4_PROPERTY_ID = "352643189"
BQ_PROJECT = "skycop-website-analytics"
BQ_DATASET = "Historical_GA4"
CREDENTIALS_PATH = "/content/skycop-website-analytics-ee9bebd0306e.json"
DATE_START = "2025-03-18"
DATE_END = "2025-04-16"

# === CLIENT SETUP ===
credentials = service_account.Credentials.from_service_account_file(CREDENTIALS_PATH)
ga_client = BetaAnalyticsDataClient(credentials=credentials)
bq_client = bigquery.Client(credentials=credentials, project=BQ_PROJECT)

# === INFERRED SCHEMA TYPES ===
TYPE_MAP = {
    "date": "DATE",
    "dateHourMinute": "DATETIME",
    "pagePath": "STRING",
    "pageTitle": "STRING",
    "eventName": "STRING",
    "firstUserCampaignName": "STRING",
    "firstUserSourceMedium": "STRING",
    "fullPageUrl": "STRING",
    "pageLocation": "STRING",
    "sourceMedium": "STRING",
    "campaignName": "STRING",
    "firstUserDefaultChannelGroup": "STRING",
    "firstUserSource": "STRING",
    "firstUserMedium": "STRING",
    "sessionDefaultChannelGroup": "STRING",
    "transactionId": "STRING",
    "country": "STRING",
    "region": "STRING",
    "city": "STRING",
    "language": "STRING",
    "userGender": "STRING",
    "userAgeBracket": "STRING",
    "platform": "STRING",
    "deviceCategory": "STRING",
    "operatingSystem": "STRING",
    "browser": "STRING",
    # Metrics
    "sessions": "INTEGER",
    "engagedSessions": "INTEGER",
    "totalUsers": "INTEGER",
    "transactions": "INTEGER",
    "eventCount": "INTEGER",
    "eventValue": "FLOAT",
    "conversions": "INTEGER",
    "activeUsers": "INTEGER",
    "bounceRate": "FLOAT",
    "averageRevenuePerUser": "FLOAT",
    "averagePurchaseRevenue": "FLOAT",
    "ecommercePurchases": "INTEGER",
    "screenPageViews": "INTEGER",
    "userEngagementDuration": "FLOAT"
}

# === REPORTS ===
REPORTS = {
    "traffic_acquisition": {
        "dimensions": ["date", "sessionDefaultChannelGroup", "source", "medium", "campaignName"],
        "metrics": ["sessions", "engagedSessions", "bounceRate"]
    },
    "user_acquisition": {
        "dimensions": ["date", "firstUserDefaultChannelGroup", "firstUserSource", "firstUserMedium"],
        "metrics": ["totalUsers", "engagedSessions", "transactions"]
    },
    "engagement_overview": {
        "dimensions": ["date", "pagePath", "pageTitle"],
        "metrics": ["screenPageViews", "userEngagementDuration", "engagedSessions", "transactions"]
    },
    "events": {
        "dimensions": ["date", "eventName"],
        "metrics": ["eventCount", "eventValue"]
    },
    "conversions": {
        "dimensions": ["date", "eventName"],
        "metrics": ["conversions", "totalUsers", "transactions"]
    },
    "General": {
        "dimensions": ["date", "eventName", "firstUserCampaignName", "firstUserSourceMedium", "fullPageUrl", "pageLocation", "sourceMedium"],
        "metrics": ["activeUsers", "bounceRate", "averageRevenuePerUser", "eventValue", "sessions"]
    },
    "General_revenue": {
        "dimensions": ["date", "eventName", "firstUserCampaignName", "firstUserSourceMedium", "pageLocation"],
        "metrics": ["activeUsers", "bounceRate", "averagePurchaseRevenue", "eventValue", "sessions", "ecommercePurchases"]
    },
    "claims": {
        "dimensions": ["date", "eventName", "firstUserCampaignName", "firstUserSourceMedium", "pageLocation"],
        "metrics": ["sessions", "transactions"]
    },
    "claims_sessions_users_ids": {
        "dimensions": ["dateHourMinute", "eventName", "firstUserCampaignId", "sessionCampaignId", "pageLocation", "transactionId", "country", "region", "city"],
        "metrics": ["activeUsers"]
    },
    "demographics": {
        "dimensions": ["date", "country", "region", "city", "language", "userGender", "userAgeBracket"],
        "metrics": ["totalUsers", "sessions", "transactions"]
    },
    "tech": {
        "dimensions": ["date", "platform", "deviceCategory", "operatingSystem", "browser"],
        "metrics": ["totalUsers", "sessions", "transactions"]
    }
}

# === EXPORT FUNCTION ===
def export_report_to_bq(report_name, dimensions, metrics):
    request = RunReportRequest(
        property=f"properties/{GA4_PROPERTY_ID}",
        dimensions=[Dimension(name=d) for d in dimensions],
        metrics=[Metric(name=m) for m in metrics],
        date_ranges=[DateRange(start_date=DATE_START, end_date=DATE_END)]
    )

    response = ga_client.run_report(request)

    # Collect and convert data
    data = []
    for row in response.rows:
        record = {}
        for i, dim in enumerate(dimensions):
            val = row.dimension_values[i].value
            if dim == "date":
                record[dim] = pd.to_datetime(val).date()
            elif dim == "dateHourMinute":
                record[dim] = pd.to_datetime(val, format="%Y%m%d%H%M")
            else:
                record[dim] = val

        for j, met in enumerate(metrics):
            val = row.metric_values[j].value
            if TYPE_MAP.get(met) == "INTEGER":
                record[met] = int(val)
            elif TYPE_MAP.get(met) == "FLOAT":
                record[met] = float(val)
            else:
                record[met] = val

        data.append(record)

    df = pd.DataFrame(data)

    # Define BigQuery schema
    schema = []
    for col in dimensions + metrics:
        bq_type = TYPE_MAP.get(col, "STRING")
        schema.append(bigquery.SchemaField(col, bq_type))

    table_id = f"{BQ_PROJECT}.{BQ_DATASET}.{report_name}"
    job_config = bigquery.LoadJobConfig(schema=schema, write_disposition="WRITE_TRUNCATE")

    # Upload to BigQuery
    job = bq_client.load_table_from_dataframe(df, table_id, job_config=job_config)
    job.result()
    print(f"✅ Exported: {report_name}")

# === RUN EXPORTS ===
for report_name, config in REPORTS.items():
    try:
        export_report_to_bq(report_name, config["dimensions"], config["metrics"])
    except Exception as e:
        print(f"❌ Failed to export {report_name}: {e}")
```
