```python
def get_facebook_client(my_app_id, my_app_secret, my_access_token, ad_account_id):
    FacebookAdsApi.init(my_app_id, my_app_secret, my_access_token, api_version='v20.0')
    facebook_account = AdAccount('act_'+ str(ad_account_id))
    return facebook_account

def bq_job_config(schema):
    '''
    Creates a config for uploading to Google BigQuery
    :schema: Google BigQuery table schema
    '''

    bq_config = bigquery.LoadJobConfig()
    bq_config.write_disposition = 'WRITE_APPEND'
    bq_config.source_format = "NEWLINE_DELIMITED_JSON"
    bq_config.schema = schema
    bq_config.autodetect = True
    print('Job config was prepared')

    return bq_config

def insert_json_bq(bigquery_client, table_id, dataset_id, project_id, data, schema):
    '''
    Uploads data to Google BigQuery table
    :bigquery_client: Google BigQuery Client
    :table_id: table identifier
    :dataset_id: dataset identifier
    :project_id: project ID
    :data: ND JSON that has to be loaded
    :schema: Google BigQuery table schema
    '''

    job_config = bq_job_config(schema)

    table_ref = f"{project_id}.{dataset_id}.{table_id}"

    table = bigquery_client.get_table(table_ref)

    resp = bigquery_client.load_table_from_json(
        json_rows=data,
        destination=table_ref,
        job_config=job_config
    )

    resp.result()

    print('Data was uploaded to the table %s', table.table_id)

def exist_dataset_table(bigquery_client, table_id, dataset_id, project_id, schema, clustering_fields=None):
    """Get the existing dataset and the table in the current project.
        If they don't exist: create the dataset and the table in the current project.

    Args:
        bigquery_client: BigQuery Client.
        table_id (str): The table ID of BigQuery.
        dataset_id (str): The dataset ID of BigQuery.
        project_id (str): The project name of Google Cloud account.
        schema: The schema of the target BigQuery table.
        clustering_fields: Fields for clustering.

    Returns:
        str
    """

    try:
        dataset_ref = "{}.{}".format(project_id, dataset_id)
        bigquery_client.get_dataset(dataset_ref)

    except NotFound:
        dataset_ref = "{}.{}".format(project_id, dataset_id)
        dataset = bigquery.Dataset(dataset_ref)
        dataset.location = "US"
        dataset = bigquery_client.create_dataset(dataset)
        print("Created dataset {}.{}".format(bigquery_client.project, dataset.dataset_id))

    try:
        table_ref = "{}.{}.{}".format(project_id, dataset_id, table_id)
        bigquery_client.get_table(table_ref)

    except NotFound:

        table_ref = "{}.{}.{}".format(project_id, dataset_id, table_id)

        table = bigquery.Table(table_ref, schema=schema)



        if clustering_fields is not None:
            table.clustering_fields = clustering_fields

        table = bigquery_client.create_table(table)
        print("Created table {}.{}.{}".format(table.project, table.dataset_id, table.table_id))

    return "ok"

schema_fb = [
    bigquery.SchemaField("date_start", "DATE", mode="NULLABLE"),
    bigquery.SchemaField("date_stop", "DATE", mode="NULLABLE"),
    bigquery.SchemaField("campaign_id", "INTEGER", mode="REQUIRED"),
    bigquery.SchemaField("campaign_name", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("account_id", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("spend", "FLOAT", mode="REQUIRED"),
    bigquery.SchemaField("clicks", "INTEGER", mode="REQUIRED"),
    bigquery.SchemaField("outbound_clicks", "RECORD", mode="REPEATED", fields=[
        bigquery.SchemaField("action_type", "STRING", mode="NULLABLE"),
        bigquery.SchemaField("value", "FLOAT", mode="NULLABLE")
    ]),
    bigquery.SchemaField("impressions", "INTEGER", mode="REQUIRED"),
    bigquery.SchemaField("actions", "RECORD", mode="REPEATED", fields=[
        bigquery.SchemaField("action_type", "STRING", mode="NULLABLE"),
        bigquery.SchemaField("value", "FLOAT", mode="NULLABLE")
    ]),
    bigquery.SchemaField("link_url_asset", "RECORD", mode="REPEATED", fields=[
        bigquery.SchemaField("website_url", "STRING", mode="NULLABLE"),
        bigquery.SchemaField("id", "INTEGER", mode="NULLABLE")
    ]),
    bigquery.SchemaField("conversions", "RECORD", mode="REPEATED", fields=[
        bigquery.SchemaField("action_type", "STRING", mode="NULLABLE"),
        bigquery.SchemaField("value", "STRING", mode="NULLABLE")
    ])
]
def get_facebook_insights(facebook_account, since, until):
    try:
        job = facebook_account.get_insights(
            fields=[
                AdsInsights.Field.date_start,
                AdsInsights.Field.date_stop,
                AdsInsights.Field.account_id,
                AdsInsights.Field.campaign_id,
                AdsInsights.Field.campaign_name,
                AdsInsights.Field.spend,
                AdsInsights.Field.impressions,
                AdsInsights.Field.clicks,
                AdsInsights.Field.outbound_clicks,
                AdsInsights.Field.actions,
                AdsInsights.Field.conversions
            ],
            params={
                'level': 'campaign',
                'time_range': {
                    'since': since,
                    'until': until
                },
                # 'breakdowns': ['link_url_asset'],
                'time_increment': 1
            },
            is_async=True
        )

        while True:
            job = job.api_get()
            time.sleep(1)
            if job[AdReportRun.Field.async_status] == 'Job Completed':
                print("Async job is done")
                insights = job.get_result(params={"limit": 1000})
                break

        if len(insights) == 0:
            print(f"⏹No data from {since} to {until}")
            return 'ok'

        return insights

    except Exception as e:
        print(f"Got error: {e}")
        raise
```

```python
import time
from datetime import datetime, timedelta

def get_data_and_load_to_bq(event, start_date, end_date, schema_fb):
    if 'startdate' in event['attributes']:
        date_from = event['attributes']['startdate']
    else:
        date_from = start_date

    if 'enddate' in event['attributes']:
        date_to = event['attributes']['enddate']
    else:
        date_to = end_date

    app_id = event['attributes']['MY_APP_ID']
    app_secret = event['attributes']['MY_APP_SECRET']
    long_lived_access_token = event['attributes']['MY_LONG_LIVED_ACCESS_TOKEN']
    ad_account_id = event['attributes']['AD_ACCOUNT_ID']
    dataset_id = event['attributes']['dataset_id']
    project_id = event['attributes']['project_id']
    table_id = event['attributes']['table_id']

    facebook_account = get_facebook_client(app_id, app_secret, long_lived_access_token, ad_account_id)

    length = (datetime.strptime(date_to, '%Y-%m-%d') - datetime.strptime(date_from, '%Y-%m-%d')).days

    for i in range(0, length + 1):
        start_date = (datetime.strptime(date_from, '%Y-%m-%d') + timedelta(i)).strftime('%Y-%m-%d')
        end_date = (datetime.strptime(date_from, '%Y-%m-%d') + timedelta(i)).strftime('%Y-%m-%d')

        retries = 3  # Number of retries for rate limit errors
        for attempt in range(retries):
            try:
                # Fetch insights
                insights = get_facebook_insights(facebook_account, start_date, end_date)

                if insights == 'ok':
                    return 'ok'

                fb_source = []
                for item in insights:
                    actions = []
                    if 'actions' in item:
                        for value in item['actions']:
                            actions.append({
                                'action_type': value.get('action_type', 'N/A'),
                                'value': float(value.get('value', 0))
                            })

                    outbound_clicks = []
                    if 'outbound_clicks' in item:
                        for value in item['outbound_clicks']:
                            outbound_clicks.append({
                                'action_type': value.get('action_type', 'N/A'),
                                'value': float(value.get('value', 0))
                            })

                    conversions = []
                    if 'conversions' in item:
                        for value in item['conversions']:
                            conversions.append({
                                'action_type': value.get('action_type', 'N/A'),
                                'value': float(value.get('value', 0))
                            })

                    link_url_asset = []
                    if 'link_url_asset' in item:
                        if isinstance(item['link_url_asset'], dict):
                            link_url_asset.append({
                                'website_url': item['link_url_asset'].get('website_url', 'N/A'),
                                'id': item['link_url_asset'].get('id', 'N/A')
                            })

                    fb_source.append({
                        'actions': actions,
                        'campaign_name': item.get('campaign_name', 'N/A'),
                        'date_start': item.get('date_start', 'N/A'),
                        'date_stop': item.get('date_stop', 'N/A'),
                        'spend': float(item.get('spend', 0.0)),
                        'clicks': int(item.get('clicks', 0)),
                        'outbound_clicks': outbound_clicks,
                        'impressions': int(item.get('impressions', 0)),
                        'account_id': item.get('account_id', 'N/A'),
                        'campaign_id': item.get('campaign_id', 'N/A'),
                        'link_url_asset': link_url_asset,
                        'conversions': conversions
                    })

                print(f'✅Fetched actions data from {start_date} to {end_date}')
                if exist_dataset_table(bigquery_client, table_id, dataset_id, project_id, schema_fb):
                    insert_json_bq(bigquery_client, table_id, dataset_id, project_id, fb_source, schema_fb)
                    print(f'✅ Data inserted into BigQuery for {start_date}')
                break  # Exit retry loop if successful

            except Exception as e:
                print(f"Attempt {attempt + 1} failed for {start_date}: {e}")
                if "too many calls" in str(e).lower():
                    time.sleep(60)  # Wait before retrying
                else:
                    raise e  # Raise other errors immediately
```
